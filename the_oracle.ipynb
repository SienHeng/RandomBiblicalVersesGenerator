Dataset Introduction

The dataset I used is openbible.com, a platform that digitizes the Bible in various downloadable formats, such as PDF, DOCX, and, of course, the format I used for data processing, TXT. Each dataset, which is also a file, contains the entire Bible and is organized as follows: each line represents a biblical verse, and the book name, chapter, and verse number index the start of each line. And one good thing about this format, also the reason why I choose this dataset, is that although the text is directly from the Bible, it is openbible.com that decided to have all titles of each book at the beginning of them removed, making it easier to do data processing because we do not have to pay attention to removing them, other than eliminating punctuation and the heading of each line.

```python
import requests


def get_kjv() -> str:
    response = requests.get('https://openbible.com/textfiles/kjv.txt')
    return response.text
```

```python
import string

def remove_punctuation(s) -> str:
    ret = []
    for c in s:
        if c not in string.punctuation:
            ret.append(c)
    return ''.join(ret)


def process_kjv(kjv) -> tuple[list[str], list[int]]:
    words = []
    line_lengths = []
    lines = kjv.split('\n')
    for line in lines:
        line = remove_punctuation(line)
        line_words = line.split()[2:]
        words.extend(line_words)
        line_lengths.append(len(line_words))
    return words, line_lengths
```

```python
import random


def gen_word(words: list[str]) -> str:
    return random.choice(words)


def gen_length(line_lengths: list[int]) -> int:
    return random.choice(line_lengths)


def gen_verse(words: list[str], line_lengths: list[int]) -> str:
    length = gen_length(line_lengths)
    verse = []
    for _ in range(length):
        verse.append(gen_word(words))
    verse = ' '.join(verse) + '.'
    verse = verse[0].upper() + verse[1:]
    return verse
```

```python
def main():
    print("--- The Oracle ---")
    print("Getting KJV Bible ready...")
    try:
      kjv = get_kjv()
    except:
      print("Failed to fetch the KJV file, there is likely some issue with Colab if you ever have this error reported, wait for a while and try again.")
      return
    words, line_lengths = process_kjv(kjv)
    print("Press ENTER to get a new verse, press Ctrl-C to exit.")
    while True:
        try:
            cmd = input().lower()
        except KeyboardInterrupt:
            break
        if cmd == 'quit':
            break
        verse = gen_verse(words, line_lengths)
        print(verse)
    print('Exit.')


if __name__ == '__main__':
    main()
```
